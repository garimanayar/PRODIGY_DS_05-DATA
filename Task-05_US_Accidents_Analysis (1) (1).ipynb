```python
# 📊 Step 2: Load the data
import pandas as pd

df = pd.read_csv("US_Accidents_March23.csv")
df.shape
```

```python
# 👀 Step 3: View first few rows
df.head()
```

```python
# 🧾 Step 4: Check for missing data
df.info()
df.isnull().sum()
```

```python
# ⏰ Step 5: Analyze accidents by hour
import matplotlib.pyplot as plt
import seaborn as sns

df['Start_Time'] = pd.to_datetime(df['Start_Time'])
df['Hour'] = df['Start_Time'].dt.hour

sns.histplot(df['Hour'], bins=24)
plt.title('Accidents by Hour')
plt.xlabel('Hour of the Day')
plt.ylabel('Number of Accidents')
plt.show()
```

```python
# 🌧️ Step 6: Analyze by Weather Condition
sns.countplot(y='Weather_Condition', data=df, order=df['Weather_Condition'].value_counts().iloc[:10].index)
plt.title('Top 10 Weather Conditions')
plt.xlabel('Number of Accidents')
plt.ylabel('Weather Condition')
plt.show()
```

```python
# 🌍 Step 7: Accidents by State
top_states = df['State'].value_counts().head(10)
top_states.plot(kind='bar')
plt.title('Top 10 States with Most Accidents')
plt.xlabel('State')
plt.ylabel('Number of Accidents')
plt.show()
```

```python
# 🏙️ Step 8: Accidents by City
top_cities = df['City'].value_counts().head(10)
sns.barplot(x=top_cities.values, y=top_cities.index)
plt.title('Top 10 Cities with Most Accidents')
plt.xlabel('Number of Accidents')
plt.ylabel('City')
plt.show()
```

## ✅ Summary
- The dataset contains millions of accident records across the US.
- Most accidents happen during rush hours: 7–9 AM and 4–6 PM.
- Clear weather is the most common condition.
- California, Florida, and Texas are top accident-prone states.
- Cities like Los Angeles, Houston, and Charlotte see many accidents.
